# Portfolio Optimization Parameters and functions

```python
delta= 0.5
lamb= np.matrix.transpose(np.asmatrix(np.array([0.5,0.5])))

Gamma= M + delta /(1-delta) * np.matrix.transpose(Q) @ rho @ np.matrix.transpose(lamb)
C= 0.5 * delta / (1-delta)**2 * lamb @ np.matrix.transpose(lamb)
Lambda= np.matrix.transpose(Q) @ ((1-delta)* np.eye(2) + delta * rho @ np.matrix.transpose(rho)) @ Q
```

```python
def split(array, nrows, ncols):
        """Split a matrix into sub-matrices."""
        r, h = array.shape
        return  (array.reshape(h//nrows, nrows, -1, ncols)
                    .swapaxes(1, 2)
                    .reshape(-1, nrows, ncols))

def Lamb(tau):
    lambda_11= Gamma
    lambda_12= -2 * Lambda
    lambda_21= C
    lambda_22= - np.matrix.transpose(Gamma)
    lambd= np.asmatrix(np.block([[lambda_11, lambda_12], [lambda_21,lambda_22]]))
    #output= np.asmatrix(scipy.linalg.expm(tau * lambd))
    output= scipy.linalg.expm(tau * lambd)
    return output

def B(tau):
    F_11, F_12, F_21, F_22=  split(Lamb(tau), 2, 2)
    return np.linalg.inv(F_22) @ F_21

#Optimal Portolfio for Wishart
def pi_W(tau):
    return lamb/(1-delta) + 2 * B(tau) * np.matrix.transpose(Q) @ rho

#Optimal Portolfio for Shifted Wishart
def pi_SW(tau, Sigma_SW):
    return lamb/(1-delta) + 2 * B(tau) * np.matrix.transpose(Q) @ rho - 2 * np.linalg.inv(Sigma_SW) @ sighat @ B(tau) @ np.matrix.transpose(Q) @ rho

#Optimal Portolfio for Markov Wishart
def pi_MW(tau, Sigma_MW):
    return lamb/(1-delta) + 2 * B(tau) * np.matrix.transpose(Q) @ rho - 2 * np.linalg.inv(Sigma_MW) @ sighat @ B(tau) @ np.matrix.transpose(Q) @ rho # There is an error in sighat term
```

# Generating SW and MW matrix processes

```python
#Wishart Simulation using Euler-Maruyama Simulation
Sigma_EM = np.zeros((N, paths, 2, 2))
Sigma_EM[0,:]= Sigma_0
W = ss.norm.rvs(loc=0, scale=1, size=(N - 1, paths, 2 , 2))
for i in range(0,N-1):
    for j in range(paths):
        sqrt_Sigma_EM= np.real(scipy.linalg.sqrtm(Sigma_EM[i,j]))
        drift=  (beta * (np.matrix.transpose(Q) @ Q) + M @ Sigma_EM[i,j] + Sigma_EM[i,j] @ np.matrix.transpose(M))* dt
        diffusion= np.sqrt(dt) * ( (sqrt_Sigma_EM @ W[i,j]) @ Q +  (np.matrix.transpose(Q) @ np.matrix.transpose(W[i,j])) @ sqrt_Sigma_EM)
        Sigma_EM[i+1,j] = Sigma_EM[i, j] + drift + diffusion


# Ensure all values are symmetric and positive semi-definite
for i in range(N):
    for j in range(paths):
        Sigma_EM[i, j] = (Sigma_EM[i, j] + np.transpose(Sigma_EM[i, j])) / 2  # Symmetrize to avoid numerical issues
        if np.any(np.linalg.eigvals(Sigma_EM[i, j]) <= 0):
            Sigma_EM[i, j] += -1.1 * np.min(np.linalg.eigvals(Sigma_EM[i, j])) * np.eye(2)  # Make positive definite

Sigma_WP= Sigma_EM
```

```python
# Simulation of Shifted Wishart Process
Sigma_SW= np.zeros((N, paths, 2, 2))
for i in range(N):
    Sigma_SW[i,:]= sighat + Sigma_WP[i,:]
```

```python
# Simulation of MArkov Modulated Shifted Wishart Process
#p=0.5
class MarkovRegimeChange:
  """
  Simulates a Markov regime change process with two regimes.

  Attributes:
    transition_matrix (np.array): Transition probability matrix.
    initial_state (int): Initial state of the process (0 or 1).
    num_steps (int): Number of steps to simulate.
  """

  def __init__(self, transition_matrix, initial_state, num_steps):
    """
    Initializes the Markov regime change process.

    Args:
      transition_matrix (np.array): 2x2 transition probability matrix.
      initial_state (int): Initial state of the process (0 or 1).
      num_steps (int): Number of steps to simulate.
    """
    self.transition_matrix = transition_matrix
    self.initial_state = initial_state
    self.num_steps = num_steps

  def simulate(self):
    """
    Simulates the Markov regime change process.

    Returns:
      np.array: Array of states for each step in the simulation.
    """
    states = np.zeros(self.num_steps, dtype=int)
    states[0] = self.initial_state

    for i in range(1, self.num_steps):
      current_state = states[i-1]
      next_state = np.random.choice(2, p=self.transition_matrix[current_state])
      states[i] = next_state

    return states

# Example usage
#transition_matrix = np.array([[p, 1-p], [1-p, p]])  # Example transition probabilities
transition_matrix = np.array([[1- lambda_01, lambda_01], [lambda_10, 1- lambda_10]])  # Example transition probabilities
#initial_state = 0
num_steps = N

States=[]
for j in range(M_paths):
  initial_state= np.random.choice([0, 1], p=[p, 1-p])
  model = MarkovRegimeChange(transition_matrix, initial_state, num_steps)
  states = model.simulate()
  States.append(states)

States= np.array(States)

Sigma_MW= np.zeros((N, paths, 2, 2))
for path in range(paths):
    for i in range(N-1):
      Sigma_MW[i,path,:]= States[path,i] * sighat + Sigma_WP[i,path,:]
```

# Optimal Portfolio paths

```python
# Optimal Portfolio under Wishart Process
OP_W=np.zeros((paths,N-1,2,1))
for path in range(paths):
    for i in range(N-1):
        tau=1-i/N #T-t
        OP_W[path,i,:]= pi_W(tau)

# Optimal Portfolio under Shifted Wishart Process
OP_SW=np.zeros((paths,N-1,2,1))
for path in range(paths):
    for i in range(N-1):
        tau=1-i/N #T-t
        OP_SW[path,i,:]= pi_SW(tau,Sigma_SW[i,path,:] )

# Optimal Portfolio under Markov Wishart Process
OP_MW=np.zeros((paths,N-1,2,1))
for path in range(paths):
    for i in range(N-1):
        tau=1- i/N #T-t
        OP_MW[path,i,:]= pi_MW(tau,Sigma_MW[i,path,:] )
```
